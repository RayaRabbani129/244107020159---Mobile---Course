```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('wbc (1).csv')
df.head()
```

```python
df.info()
```

**1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.**

```python
unusable_columns = ['id', 'Unnamed: 32']
df = df.drop(unusable_columns, axis=1)
df.head()
```

**2. Lakukan proses encoding pada kolom "diagnosis".**

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
df.head()
```

**3. Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik.**

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.drop('diagnosis')
scaler = StandardScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
df.head()
```

**4. Lakukan proses stratified split data untuk membuat data latih dan data uji dengan rasio 80:20.**

```python
usable_train, usable_test = train_test_split(df, test_size=0.2, random_state=42, stratify=df['diagnosis'])

print(f'Jumlah label data asli:\n{df.diagnosis.value_counts()}')
print(f'\nJumlah label data train:\n{usable_train.diagnosis.value_counts()}')
print(f'\nJumlah label data test:\n{usable_test.diagnosis.value_counts()}')
```
